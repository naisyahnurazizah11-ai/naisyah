```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# 1. Membaca dataset
df = pd.read_csv('dataset_fiskal_97_hipoFull.csv')

# 2. Filter data khusus Kota Malang dan urutkan berdasarkan tahun
df_malang = (
    df[df['Kota'] == 'Kota Malang'].sort_values('Tahun').reset_index(drop=True)
)

# Pengaturan gaya tampilan grafik
sns.set_theme(style='whitegrid')
plt.rcParams['font.sans-serif'] = 'DejaVu Sans'
plt.rcParams['figure.dpi'] = 150

```

```python
# VISUALISASI 1: Tren Indikator Makro Sosial-Ekonomi

fig, ax1 = plt.subplots(figsize=(10, 5))

color1, color2, color3, color4 = '#1f77b4', '#2ca02c', '#d62728', '#ff7f0e'

ax1.set_xlabel('Tahun', fontsize=11, fontweight='bold')
ax1.set_ylabel('Persentase / Skor (%)', fontsize=11, fontweight='bold')

l1 = ax1.plot(
    df_malang['Tahun'],
    df_malang['IPM'],
    marker='o',
    linewidth=2.5,
    color=color1,
    label='IPM (Indeks Pembangunan Manusia)',
)[0]
l2 = ax1.plot(
    df_malang['Tahun'],
    df_malang['PE'],
    marker='s',
    linewidth=2,
    linestyle='--',
    color=color2,
    label='PE (Pertumbuhan Ekonomi %)',
)[0]
l3 = ax1.plot(
    df_malang['Tahun'],
    df_malang['KEMISKINAN'],
    marker='^',
    linewidth=2,
    linestyle='-.',
    color=color3,
    label='Tingkat Kemiskinan (%)',
)[0]
l4 = ax1.plot(
    df_malang['Tahun'],
    df_malang['PENGANGGURAN'],
    marker='d',
    linewidth=2,
    linestyle=':',
    color=color4,
    label='Tingkat Pengangguran (%)',
)[0]

ax1.set_xticks(df_malang['Tahun'])
ax1.set_title(
    'Visualisasi 1: Tren Indikator Makro Sosial-Ekonomi Kota Malang'
    ' (2014-2024)',
    fontsize=12,
    fontweight='bold',
    pad=15,
)
ax1.legend(
    handles=[l1, l2, l3, l4],
    loc='center left',
    bbox_to_anchor=(1.02, 0.5),
    frameon=True,
)
plt.tight_layout()
plt.savefig('visualisasi_1_indikator_makro.png', bbox_inches='tight')
plt.show()
```

```python
# VISUALISASI 2: Struktur & Sumber Pendapatan Daerah

df_malang['Total_DBH'] = (
    df_malang['DBH_PAJAK'] + df_malang['DBH_SDA'] + df_malang['DBH_Lainnya']
)
df_malang['Total_DAU'] = df_malang['DAU_Grant'] + df_malang['DAU_Earmark']
df_malang['Total_DAK'] = df_malang['DAK_Fisik'] + df_malang['DAK_Nonfisik']

fig, ax = plt.subplots(figsize=(10, 5.5))
years = df_malang['Tahun']
bar_width = 0.65

p1 = ax.bar(
    years,
    df_malang['PAD'],
    bar_width,
    label='PAD (Pendapatan Asli Daerah)',
    color='#2b5c8f',
)
p2 = ax.bar(
    years,
    df_malang['Total_DAU'],
    bar_width,
    bottom=df_malang['PAD'],
    label='DAU (Grant + Earmark)',
    color='#4682b4',
)
p3 = ax.bar(
    years,
    df_malang['Total_DBH'],
    bar_width,
    bottom=df_malang['PAD'] + df_malang['Total_DAU'],
    label='DBH (Pajak, SDA, Lainnya)',
    color='#e07a5f',
)
p4 = ax.bar(
    years,
    df_malang['Total_DAK'],
    bar_width,
    bottom=df_malang['PAD']
    + df_malang['Total_DAU']
    + df_malang['Total_DBH'],
    label='DAK (Fisik + Nonfisik)',
    color='#81b29a',
)

ax.set_xlabel('Tahun', fontsize=11, fontweight='bold')
ax.set_ylabel('Nilai (Miliar / Satuan Data)', fontsize=11, fontweight='bold')
ax.set_xticks(years)
ax.set_title(
    'Visualisasi 2: Stuktur & Sumber Pendapatan Daerah Kota Malang (2014-2024)',
    fontsize=12,
    fontweight='bold',
    pad=15,
)
ax.legend(loc='upper left', frameon=True)
plt.tight_layout()
plt.savefig('visualisasi_2_pendapatan.png', bbox_inches='tight')
plt.show()
```

```python
# VISUALISASI 3: Perkembangan Belanja Modal per Sektor Utama

fig, ax = plt.subplots(figsize=(11, 5.5))

sektor_cols = {
    'bm_pendidikan': 'Pendidikan',
    'bm_kesehatan': 'Kesehatan',
    'bm_infrastruktur': 'Infrastruktur',
    'bm_perumahan': 'Perumahan',
    'bm_lingkungan': 'Lingkungan',
    'bm_ekonomi': 'Ekonomi',
    'bm_perlindungan_sosial': 'Perlindungan Sosial',
}

colors = sns.color_palette('Set2', len(sektor_cols))

for (col, label_name), color in zip(sektor_cols.items(), colors):
  ax.plot(
      df_malang['Tahun'],
      df_malang[col],
      marker='o',
      linewidth=2,
      label=label_name,
      color=color,
  )

ax.set_xlabel('Tahun', fontsize=11, fontweight='bold')
ax.set_ylabel('Alokasi Belanja Modal', fontsize=11, fontweight='bold')
ax.set_xticks(df_malang['Tahun'])
ax.set_title(
    'Visualisasi 3: Perkembangan Belanja Modal per Sektor Utama Kota Malang'
    ' (2014-2024)',
    fontsize=12,
    fontweight='bold',
    pad=15,
)
ax.legend(loc='center left', bbox_to_anchor=(1.02, 0.5), frameon=True)
plt.tight_layout()
plt.savefig('visualisasi_3_belanja_modal.png', bbox_inches='tight')
plt.show()
```

```python
# VISUALISASI 4: Matriks Korelasi Fiskal vs Indikator Makro

fig, ax = plt.subplots(figsize=(9, 7))

corr_cols = [
    'PAD',
    'Total_DAU',
    'Total_DAK',
    'bm_pendidikan',
    'bm_kesehatan',
    'bm_infrastruktur',
    'IPM',
    'PE',
    'KEMISKINAN',
    'PENGANGGURAN',
]
labels_clean = [
    'PAD',
    'DAU Total',
    'DAK Total',
    'BM Pendidikan',
    'BM Kesehatan',
    'BM Infrastruktur',
    'IPM',
    'PE (%)',
    'Kemiskinan (%)',
    'Pengangguran (%)',
]

corr_matrix = df_malang[corr_cols].corr()

sns.heatmap(
    corr_matrix,
    annot=True,
    fmt='.2f',
    cmap='coolwarm',
    vmin=-1,
    vmax=1,
    xticklabels=labels_clean,
    yticklabels=labels_clean,
    ax=ax,
    cbar_kws={'label': 'Koefisien Korelasi'},
)

ax.set_title(
    'Visualisasi 4: Matriks Korelasi Fiskal vs Indikator Makro Kota Malang',
    fontsize=12,
    fontweight='bold',
    pad=15,
)
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.savefig('visualisasi_4_korelasi.png', bbox_inches='tight')
plt.show()
```
